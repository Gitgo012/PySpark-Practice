## This notebook has PySpark code defining how to write to different volumes/unity catalog in different formats and as tables, how to partition the df and store it as a table if required and the different writing modes

```python
df=spark.read.table("samples.accuweather.forecast_daily_calendar_imperial")
df.display()
```

```python
df.write.format("csv").save("/Volumes/sales_catalogue_testing/sales_data_practice_schema/badrecordspath_volume/write_test1") 
```

Writing a table in Unity Catalog follows a specific format:
`catalog_name.schema_name.table_name`

```python
df.write.saveAsTable("sales_catalogue_testing.sales_data_practice_schema.write_test1")
```

```python
df.write.mode("append").format("csv").save("/Volumes/sales_catalogue_testing/sales_data_practice_schema/badrecordspath_volume/write_test1")
```

- Generally, partitioning is considered when data is large (often several GBs or more). Very small partitions can introduce unnecessary overhead.
- Partitioning improves query performance by enabling partition pruning, where Spark reads only the relevant partitions instead of scanning the entire dataset.
- It is particularly useful for frequently filtered columns, such as year, country, or region.
- Partitioning can reduce unnecessary data scanning and improve query efficiency.
Avoid over-partitioning, as creating too many small partitions can increase metadata overhead and reduce performance.
- Choose partition columns based on common filter conditions and data distribution.

```python
from pyspark.sql.functions import year

df.write.mode("overwrite").partitionBy("country_code").save("/Volumes/sales_catalogue_testing/sales_data_practice_schema/badrecordspath_volume/write_test2")
```

- so basically table remains same with all the columns intact just that records internally are grouped by the partition key

```python
df.write.mode("overwrite").partitionBy("country_code").saveAsTable("sales_catalogue_testing.sales_data_practice_schema.write_test2")
```
