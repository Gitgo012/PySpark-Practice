## This notebook will have information on how to deal with corrupted data in PySpark using different modes (Permisive, FailFast, DropMalformed & badRecordsPath).

let's start by loading a file with bad data present in it for json object. \
By default it will use PERMISSIVE mode

```python
df=spark.read.format("json").load("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_practice/sales_json/sales_corrupted.json")
df.display()
```

let's read it in FAILFAST mode

```python
df2=spark.read.format("json").\
option("mode","FAILFAST").\
load("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_practice/sales_json/sales_corrupted.json")
df2.display()
```

now let's read the data in DROPMALFORMED mode

```python
df3=spark.read.format("json").\
    option("mode","DROPMALFORMED").\
        load("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_practice/sales_json/sales_corrupted.json")
df3.display()
```

### Now we will work towards hadling csv data 

here by default we will go with using PERMISSIVE mode however this mode will not produce any _corrupt_record since we are not defining that field in schema. It will simply remove all corrupted data with null

```python
df4=spark.read.format("csv").\
    option("mode","DROPMALFORMED").\
        load("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_practice/sales_csv/sales_corrupted.csv")
df4.display()
```

here we specify the schema we want to ensure headers are present

```python
schema_str = "id INT, customer_name STRING, product STRING, quantity INT, price DOUBLE, remarks STRING, _corrupt_record STRING"

df5=spark.read.\
schema(schema_str)\
.format("csv")\
.option("mode","DROPMALFORMED")\
.load("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_practice/sales_csv/sales_corrupted.csv")

df5.display()
```

Now let's use the **badRecordsPath** option. \
Output will show only valid data records and corrupted data will be written silently to the badRecordsPath location

```python
schema_str = "id INT, customer_name STRING, product STRING, quantity INT, price DOUBLE, remarks STRING"

df6=spark.read.\
schema(schema_str)\
.format("csv")\
.option("badRecordsPath","/Volumes/sales_catalogue_testing/sales_data_practice_schema/badrecordspath_volume/badRecordsPath_directory/")\
.load("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_practice/sales_csv/sales_corrupted.csv")

display(df6)
```

here we try reading the data from the **badRecordsPath** location in a **recursiveFileLookup** manner which means Spark will go into each and every folder and then look into the data of it

```python
df7=spark.read.format("json").\
    option("recursiveFileLookup","true").\
        load("/Volumes/sales_catalogue_testing/sales_data_practice_schema/badrecordspath_volume/badRecordsPath_directory/20260926T063245/bad_records")
df7.display()
```
