## In this notebook we learn how to connect to JDBC and run queries by connecting to a DB

let's first connecting to a SQL Warehouse using JDBC using JDBC url and an expired access token and run a query on a a table

```python
my_key=dbutils.secrets.get(scope="my_secret_scope_2",key="my_key")
```

```python
from pyspark.sql import SparkSession

#JDBC connection 
jdbc_url="jdbc:databricks://dbc-c512e9d3-bf98.cloud.databricks.com:443/default;transportMode=http;ssl=1;AuthMech=3;httpPath=/sql/1.0/warehouses/d59bddd7f7b1df37;"

access_token=my_key

query="Select * from samples.bakehouse.sales_franchises Limit 100"

#reading the data
df=spark.read.\
format("jdbc").\
option("url",jdbc_url).\
option("query",query).\
option("user","token").\
option("password",access_token).\
load()

display(df)
```

another way of running the same query for select * is by just specifying the table name in .options


```python
#reading the data
df2=spark.read.\
format("jdbc").\
option("url",jdbc_url).\
option("dbtable","samples.bakehouse.sales_franchises").\
option("user","token").\
option("password",access_token).\
load()

display(df2)
```
