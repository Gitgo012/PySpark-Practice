## Handling different kind of csv data (delimiters, quotes and multi-line data)

```python
df=spark.read.csv("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_csv_practice/sales_csv/sales.csv")
df.display()
```

Another way of reading the data is by using _**spark.read.format(format).load(path)**_

```python
df=spark.read.format("csv").load("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_csv_practice/sales_csv/sales.csv")
df.display()
```

defining headers in spark using **_.option_** (by default it reads all the columns as strings)

```python
df=spark.read.\
option("header","true")\
.csv("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_csv_practice/sales_csv/sales.csv")

display(df)
```

we can make sure it infers the columns from data using **_inferschema_** along with _**.option**_

```python
df=spark.read.\
option("header","true").\
option("inferschema","true").\
csv("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_csv_practice/sales_csv/sales.csv")

df.display()
```

we can make sure that if schema is not present we can **_spark.read.schema(schema_string)_** (pass a schema string defining the schema)

```python
schema_str = "id INT, customer_name STRING, product STRING, quantity INT, price DOUBLE, remarks STRING"

df=spark.read.\
schema(schema_str)\
.csv("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_csv_practice/sales_csv/sales_no_header.csv")

display(df)
```

another way of defining schema is by using StructFields from importing pysql.sql.types the different data classes 

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType

schema = StructType([
    StructField("id", StringType(), True),
    StructField("customer_name", StringType(), True),
    StructField("my_product", StringType(), True),
    StructField("quantity", IntegerType(), True),
    StructField("price", DoubleType(), True),
    StructField("remarks", StringType(), True)
])

df=spark.read.\
schema(schema)\
.csv("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_csv_practice/sales_csv/sales_no_header.csv")

display(df)
```

now we handle delimiters in data (pipe character)

```python
df=spark.read.\
option("header","true")\
.option("inferSchema","true")\
.option("delimiter","|")\
.csv("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_csv_practice/sales_csv/sales_pipe.csv")

display(df)
```

now we handle another delimiter (quotes)

```python
df=spark.read.\
option("header","true")\
.option("inferSchema","true")\
.option("delimiter","|")\
.option("quote","'")\
.csv("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_csv_practice/sales_csv/sales_pipe_single_quote.csv")

display(df)
```

now we deal with another escape character (#)

```python
df=spark.read.\
option("header","true")\
.option("inferSchema","true")\
.option("delimiter","|")\
.option("quote","'")\
.option("escape","#")\
.csv("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_csv_practice/sales_csv/sales_pipe_single_quote_escape.csv")

display(df)
```

now we deal with delimiters + multiline data 

```python
df=spark.read.\
option("header","true")\
.option("inferSchema","true")\
.option("delimiter","|")\
.option("quote","'")\
.option("escape","#")\
.option("multiline","true")\
.csv("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_csv_practice/sales_csv/sales_pipe_single_quote_escape_multiline.csv")

display(df)
```
