## This notebook deals with Delta Table Cloning and its techniques:
1. Shallow Cloning
2. Deep Cloning

```python
from pyspark.sql import Row

# Create initial DataFrame with sample data
data = [Row(id=1, name='Alice', age=30), Row(id=2, name='Bob', age=25), Row(id=3, name='Charlie', age=35)]
df = spark.createDataFrame(data)

# Overwrite and create a managed Delta table with the DataFrame in the training.delta_demo schema
df.write.format('delta').mode('overwrite').saveAsTable('delta_table_practice.delta.people_delta_demo_3')

# Read the Delta table to verify creation and display its contents
delta_df = spark.read.table('delta_table_practice.delta.people_delta_demo_3')
display(delta_df)
```

```python
df.write.format('delta').mode('append').save('/Volumes/delta_table_practice/delta/delta_volume/delta_test_table_3')
```

```python
#written the append 5 times so rows=15
df2=spark.read.load('/Volumes/delta_table_practice/delta/delta_volume/delta_test_table_3')
display(df2)
```

```python
%sql
create or replace table delta_table_practice.delta.people_delta_target_shallow shallow clone delta_table_practice.delta.people_delta_demo_3 
```

```python
%sql
create or replace table delta_table_practice.delta.people_delta_target_shallow deep clone delta_table_practice.delta.people_delta_demo_3 
```
