## Dealing with JSON data (reading it, handling delimiters & writing it)

here we read a json data and infer its schema

```python
df=spark.read.\
option("inferSchema","true").\
json("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_practice/sales_json/sales_single_line.json")

df.display()
```

now we deal with data that is multiline and has delimters

```python
df=spark.read.\
option("inferSchema","true").\
option("multiline","true").\
json("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_practice/sales_json/sales_multi_line.json")

df.display()

#we can also remove the redundancy of writing option multiple times and use options instead like done below
# df=spark.read.\
# options(multiline="true",inferschema="true").\
# json("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_practice/sales_json/sales_multi_line.json")

# df.display()

```

now we move forward and write to out our dataframe to volume and store it in the form of a json

```python
df.write.json("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_practice/sales_json/saved_as_json")
```

we will now go forward and save our dataframe as a csv too

```python
df.write.format("csv").save("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_practice/sales_csv/saved_as_csv")
```
