## This notebook deals with Unstructured data reading in PySpark in Databricks using Binary Files option. 

each line from the text file will be displayed as a single record/row

```python
df=spark.read.format("text").load("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_practice/sales_text/")
df.display()
```

we can read a single file as a single row by using **wholeText=true** \
Since we have 2 files in sales_text folder each will be displayed as a seperate record/row

```python
df2=spark.read.format("text").\
    option("wholeText","true").\
        load("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_practice/sales_text/")
df2.display()
```

now if we want to try to read any file (audio,video,text,etc.) then we can use **binaryFile** as the file format like we specify **csv**, **json**, etc.

```python
df3=spark.read.format("binaryFile").\
    load("/Volumes/sales_catalogue_testing/sales_data_practice_schema/sales_practice/sales_text/")
df3.display() 
```

we can also see additional file informations using **.selectExpr** & **_metadata**. \
**selectExpr** creates a new dataframe due to which it needs to be stored seperately in a new df.

```python
df4=df2.selectExpr("value","_metadata.file_path","_metadata.file_size","_metadata.file_modification_time")
df4.display()
```
